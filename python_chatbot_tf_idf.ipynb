# Python Chatbot
A complete beginner-friendly chatbot using **Python, NLTK, TF-IDF, and cosine similarity**.

The chatbot learns from a small set of intents and can be extended by adding more patterns and responses.

## 2. Import libraries and define training data

```python
import random
import re
import nltk
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download('punkt', quiet=True)

intents = {
    'greeting': {
        'patterns': ['hello', 'hi', 'hey', 'good morning', 'good evening', 'how are you'],
        'responses': ['Hello! How can I help you?', 'Hi! What can I do for you?', 'Hey! Nice to meet you.']
    },
    'name': {
        'patterns': ['what is your name', 'who are you', 'tell me your name'],
        'responses': ['I am a Python chatbot.', 'You can call me PyBot.']
    },
    'python': {
        'patterns': ['what is python', 'tell me about python', 'why learn python', 'python language'],
        'responses': ['Python is a popular, high-level programming language used in web development, data science, AI, automation, and more.']
    },
    'machine_learning': {
        'patterns': ['what is machine learning', 'explain machine learning', 'tell me about ml'],
        'responses': ['Machine learning is a branch of AI in which computers learn patterns from data to make predictions or decisions.']
    },
    'help': {
        'patterns': ['help me', 'what can you do', 'how can you help me'],
        'responses': ['I can answer simple questions about Python, machine learning, my name, and greetings.']
    },
    'thanks': {
        'patterns': ['thanks', 'thank you', 'thank you so much'],
        'responses': ['You are welcome!', 'Happy to help!', 'Anytime!']
    },
    'goodbye': {
        'patterns': ['bye', 'goodbye', 'see you', 'exit', 'quit'],
        'responses': ['Goodbye!', 'See you later!', 'Have a great day!']
    }
}
```

## 3. Prepare the chatbot model

```python
sentences = []
labels = []

for intent, data in intents.items():
    for pattern in data['patterns']:
        sentences.append(pattern)
        labels.append(intent)

vectorizer = TfidfVectorizer(lowercase=True, stop_words='english', ngram_range=(1, 2))
X = vectorizer.fit_transform(sentences)

def clean_text(text):
    text = text.lower().strip()
    text = re.sub(r'[^a-z0-9\s]', '', text)
    return text

def chatbot_response(user_input, threshold=0.25):
    user_input = clean_text(user_input)
    if not user_input:
        return 'Please type something.'

    user_vector = vectorizer.transform([user_input])
    scores = cosine_similarity(user_vector, X)[0]
    best_index = scores.argmax()
    best_score = scores[best_index]

    if best_score < threshold:
        return "Sorry, I don't understand that yet. Try asking about Python, machine learning, or help."

    intent = labels[best_index]
    return random.choice(intents[intent]['responses'])
```

## 4. Test the chatbot

```python
test_questions = [
    'Hello chatbot',
    'What is Python?',
    'Explain machine learning',
    'What can you do?',
    'Thanks!'
]

for question in test_questions:
    print(f'You: {question}')
    print(f'Bot: {chatbot_response(question)}\n')
```

## 5. Run the chatbot interactively

```python
print("PyBot: Hello! Type 'bye' to exit.\n")

while True:
    user_input = input('You: ')
    response = chatbot_response(user_input)
    print('PyBot:', response)

    if clean_text(user_input) in {'bye', 'goodbye', 'exit', 'quit'}:
        break
```

### How it works
1. User input is cleaned.
2. TF-IDF converts text into numerical vectors.
3. Cosine similarity compares the input with known training phrases.
4. The closest matching intent is selected.
5. A response is randomly selected from that intent.
