## 1. Import ChatPromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate

print('ChatPromptTemplate imported successfully!')
```

## 2. Create a basic chat prompt template
Use message roles such as `system` (instructions) and `human` (user input). Variables inside `{curly_braces}` are filled in later.

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful Python tutor. Explain concepts clearly and give examples."),
    ("human", "Explain {topic} in simple language for a {level} student.")
])

print(prompt)
```

## 3. Fill the template with values

```python
messages = prompt.format_messages(topic="Python loops", level="beginner")

for message in messages:
    print(f"{message.type.upper()}: {message.content}")
```

## 4. Use `format()` to get a formatted prompt string
This is useful for inspecting the final prompt before sending it to a model.

```python
formatted_text = prompt.format(topic="lists", level="beginner")
print(formatted_text)
```

## 5. Reuse a template for different inputs

```python
topics = ["Python functions", "dictionaries", "exception handling"]

for topic in topics:
    messages = prompt.format_messages(topic=topic, level="intermediate")
    print(f"\nTopic: {topic}")
    print(messages[-1].content)
```

## 6. Add multiple messages and placeholders
A chat prompt can include system instructions, example dialogue, and a human request.

```python
conversation_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a concise assistant that helps with {domain}."),
    ("human", "What is {concept}?"),
    ("ai", "I will explain {concept} in the context of {domain}."),
    ("human", "Now give me one practical example.")
])

result = conversation_prompt.format_messages(domain="data science", concept="overfitting")
for message in result:
    print(f"{message.type}: {message.content}")
```

## 7. Validate required input variables
Check which variables the template expects, then provide them when formatting.

```python
print("Required variables:", prompt.input_variables)

# Example with all required variables supplied:
ready_messages = prompt.format_messages(topic="classes and objects", level="beginner")
print("Number of messages:", len(ready_messages))
```
