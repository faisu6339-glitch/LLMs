```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# ============================================
# Candidate Profile
# ============================================

candidate_profile = """
Python Developer

5 years of experience

Skills:
- Python
- AWS
- LangChain
- Generative AI
- Docker
"""


# ============================================
# Chain 1: Job Search
# ============================================

jobsearch_prompt = ChatPromptTemplate.from_template(
    """
    Based on the following candidate profile, generate a list
    of 10 suitable job opportunities.

    Candidate Profile:
    {candidate_profile}

    For each job provide:
    1. Job title
    2. Company
    3. Required skills
    4. Experience required
    5. Location
    6. Reason why the candidate matches
    """
)

jobsearch_chain = jobsearch_prompt | llm | StrOutputParser()


# ============================================
# Chain 2: Interview Questions
# ============================================

quesprep_prompt = ChatPromptTemplate.from_template(
    """
    Based on the following job opportunities, prepare interview
    questions.

    Job Opportunities:
    {list_jobs}

    Prepare 15 interview questions for each job.

    Include questions from:
    - Python
    - AWS
    - LangChain
    - Generative AI
    - Docker
    - Project experience
    - Scenario-based questions
    """
)

quesprep_chain = quesprep_prompt | llm | StrOutputParser()
```

## Understanding LangChain in LLM Context

LangChain is an open-source framework designed to simplify the development of applications powered by Large Language Models (LLMs). In essence, it acts as a "glue" or "orchestration layer" that allows developers to compose and chain together various components (LLMs, prompt templates, external data, tools, memory, etc.) to build more sophisticated, context-aware, and interactive LLM applications.

To understand LangChain in detail, let's break down its purpose, core components, and how it addresses the limitations of raw LLMs.

---

### The Problem LangChain Solves: Limitations of Raw LLMs

While LLMs are incredibly powerful, when used in isolation, they have several limitations for real-world applications:

1.  **Lack of Persistent Memory:** LLMs are stateless. Each API call is a fresh start. They don't remember previous interactions in a conversation.
2.  **Limited Context Window:** While improving, LLMs have a finite input token limit. They can't process extremely long documents or entire conversation histories without summarization or external retrieval.
3.  **No Access to Real-time or External Data:** LLMs are trained on historical data. They cannot access the internet, databases, or proprietary documents in real-time.
4.  **Inability to Perform Actions:** LLMs can generate text, but they cannot *act* in the world (e.g., search the web, run code, call an API, send an email).
5.  **Complexity of Orchestration:** Building complex workflows (e.g., "summarize this document, then answer questions about it using external search, and then present it in a specific format") requires significant boilerplate code to manage API calls, parse responses, and handle state.
6.  **Reproducibility & Iteration:** Crafting effective prompts is an iterative process. Managing different prompt versions, input variables, and few-shot examples can become cumbersome.

LangChain addresses these challenges by providing a structured way to combine LLMs with other computational and data sources, transforming them from mere text generators into powerful reasoning and action engines.

---

### LangChain's Core Components (The "Building Blocks")

LangChain's architecture is modular and comprises several key abstractions:

*   **Models:** These are the LLMs themselves, or other types of models (e.g., chat models, embedding models) that LangChain interacts with. It provides a unified interface for various LLMs (OpenAI, Google Gemini, Hugging Face, etc.).

*   **Prompts:** Templates for constructing inputs to language models. This includes prompt templates for zero-shot, few-shot, and chat models. It allows for dynamic input insertion and structured prompt creation.

*   **Chains:** Sequences of calls to LLMs or other utilities. Chains allow you to combine different components (e.g., a prompt template, an LLM, and an output parser) into a single, logical workflow. Examples include simple LLM chains, sequential chains, and retrieval chains.

*   **Retrieval Augmented Generation (RAG):** This is a powerful pattern where an LLM is augmented with external knowledge retrieved from a database or document store. LangChain provides tools for integrating vector stores, document loaders, and retrievers to enable LLMs to answer questions about specific, up-to-date, or proprietary data.

*   **Agents:** These are more dynamic and intelligent systems where an LLM acts as a reasoning engine to decide which tools to use and in what order, based on user input. Agents can perform complex tasks by interacting with external APIs, databases, or even other LLMs. They can have memory to maintain conversational context.

*   **Memory:** Allows chains and agents to remember previous interactions. LLMs are stateless, so memory components provide a way to persist and manage conversational history, enabling more coherent and context-aware dialogues.

*   **Tools:** Functions or APIs that agents can use to interact with the external world. Examples include search engines, calculators, code interpreters, or custom APIs. Tools enable LLMs to go beyond text generation and perform actions.

*   **Parsers:** Components that transform the raw output of an LLM into a more structured or usable format (e.g., JSON, Python objects, specific strings).

---

### Benefits of Using LangChain

*   **Modularity & Composability:** Break down complex tasks into smaller, manageable components that can be easily combined and reused.
*   **Interoperability:** Seamlessly switch between different LLM providers and models with a consistent API.
*   **Rapid Prototyping:** Reduces boilerplate code, allowing developers to quickly build and iterate on LLM applications.
*   **Context Management:** Provides robust solutions for handling long conversations and incorporating external knowledge (RAG).
*   **Actionability:** Enables LLMs to interact with the real world through tools and agents.
*   **Community & Ecosystem:** A large and active community, extensive documentation, and integrations with many popular services.
*   **Observability (via LangSmith):** LangChain is tightly integrated with LangSmith, a platform for debugging, testing, evaluating, and monitoring LLM applications built with LangChain.

---

### Challenges and Considerations

*   **Complexity:** LangChain can have a steep learning curve due to its many components and abstractions, especially for beginners.
*   **Overhead:** For very simple LLM calls, LangChain might introduce unnecessary complexity.
*   **Debugging Agents:** Tracing the decision-making process of agents can be challenging, though LangSmith significantly helps.
*   **Performance:** Chaining multiple LLM calls can lead to increased latency and cost. Careful design is needed.
*   **Rapid Evolution:** The framework is under active development, meaning APIs and best practices can change frequently.
*   **Prompt Engineering Still Matters:** While LangChain helps structure prompts, crafting effective prompts remains crucial for good LLM performance.

---

### Conclusion

LangChain is a powerful and versatile framework that empowers developers to move beyond simple LLM prompts and build sophisticated, intelligent applications. By providing a structured way to connect LLMs with data, memory, and tools, it transforms LLMs from mere text generators into dynamic, context-aware, and actionable agents capable of solving complex real-world problems. It's an essential tool for anyone serious about building production-ready LLM-powered applications.

                       ↓
                ┌──────────────┐

                ┌──────────────┐

                    ┌────────────┐

# First Complete Chain

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template(
    """
    Explain {topic}.

    Requirements:
    - Beginner friendly
    - Give definition
    - Give example
    - Give advantages
    - Give disadvantages
    """
)

chain = prompt | llm | StrOutputParser()

topic = input("Enter topic: ")

result = chain.invoke({
    "topic": topic
})

print(result)
```
