```python
# c. Reusable PromptTemplate with user input

from langchain_core.prompts import PromptTemplate

# 1. Create reusable template
prompt = PromptTemplate(
    input_variables=["topic", "level", "detail"],
    template="""
You are a helpful teacher.

Topic: {topic}
Learning Level: {level}
Explanation Type: {detail}

Instructions:
1. Explain the topic according to the selected level.
2. If the explanation type is Short Summary, give only the
   important points in simple language.
3. If the explanation type is Full Detail, explain the topic
   step by step with examples.
4. Use headings and bullet points wherever appropriate.
5. Keep the answer relevant to the selected topic.

Now explain the topic.
"""
)

# 2. Take input from the user

topic = input("Enter topic: ")

print("\nChoose your level:")
print("1. Beginner")
print("2. Intermediate")
print("3. Advanced")

level_choice = input("Enter your choice (1-3): ")

levels = {
    "1": "Beginner",
    "2": "Intermediate",
    "3": "Advanced"
}

level = levels.get(level_choice)

if level is None:
    print("Invalid level choice!")
else:
    print("\nChoose explanation type:")
    print("1. Short Summary")
    print("2. Full Detail")

    detail_choice = input("Enter your choice (1-2): ")

    details = {
        "1": "Short Summary",
        "2": "Full Detail"
    }

    detail = details.get(detail_choice)

    if detail is None:
        print("Invalid explanation choice!")
    else:
        # 3. Format the prompt
        final_prompt = prompt.format(
            topic=topic,
            level=level,
            detail=detail
        )

        # 4. Send prompt to your LLM
        response = llm.invoke(final_prompt)

        # 5. Display the result
        print("\n" + "=" * 50)
        print("TOPIC:", topic)
        print("LEVEL:", level)
        print("TYPE:", detail)
        print("=" * 50)

        print(response.content)
```
