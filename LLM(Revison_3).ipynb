# Program 3: Reusable topic explainer

```python

from langchain_core.prompts import PromptTemplate

# Step 1: Create PromptTemplate
prompt = PromptTemplate(
    input_variables=["Company", "Job_Role", "Experience"],
    template="""
    You are an expert career advisor and technical interviewer.

    Company: {Company}
    Job Role: {Job_Role}
    Experience: {Experience}

    Based on the information provided, complete the following tasks:

    1. Explain the job role in detail.
    2. List the important technical skills required for this role.
    3. Explain the responsibilities of this job role.
    4. Provide 5 important technical interview questions with answers.
    5. Suggest practical projects to build for this role.
    6. Create a step-by-step preparation roadmap.
    7. Provide tips to crack the interview at the given company.

    Make your response beginner-friendly, detailed, and well-structured.
    """
)

# Step 2: Format the prompt with actual values
result = prompt.format(
    Company="Google",
    Job_Role="Machine Learning Engineer",
    Experience="Fresher"
)

# Step 3: Print formatted prompt
print("Formatted Prompt:")
print(result)

# Step 4: Send prompt to LLM
response = llm.invoke(result)

# Step 5: Print LLM response
print("\nLLM Response:")
print(response.content)
```

# Program 1:	AI Career Advisor

```python

from langchain_core.prompts import ChatPromptTemplate

# Step 1: Create ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
        You are an expert Career Advisor and Technical Interviewer.

        Your responsibilities:
        1. Analyze the candidate's technical skills.
        2. Identify suitable career opportunities.
        3. Recommend relevant job roles.
        4. Identify skill gaps.
        5. Suggest technologies and tools to learn.
        6. Prepare the candidate for technical interviews.
        7. Create a practical career improvement roadmap.

        Give honest, professional and actionable advice.
        Do not recommend skills unrelated to the candidate's career goals.
        """
    ),
    (
        "human",
        """
        Candidate Name: {name}

        Educational Qualification: {education}

        Technical Skills: {skills}

        Experience: {experience}

        Career Goal: {career_goal}

        Preferred Job Role: {job_role}

        Based on the above information, provide:

        1. Candidate Profile Analysis.
        2. Three Suitable Job Roles.
        3. Technical Skill Gap Analysis.
        4. Recommended Technologies to Learn.
        5. Interview Preparation Strategy.
        6. A 90-Day Career Development Roadmap.
        7. Resume Improvement Suggestions.
        """
    )
])

# Step 2: Format the prompt
result = prompt.format_messages(
    name="Faisal",
    education="MCA",
    skills="Python, Machine Learning, Deep Learning, SQL, LangChain, RAG",
    experience="Fresher with internship experience",
    career_goal="Build a career in Artificial Intelligence",
    job_role="AI/ML Engineer"
)

# Step 3: Display messages
for message in result:
    print(message.type.upper())
    print(message.content)
    print("-" * 60)
```

```python
from langchain_core.output_parsers import StrOutputParser

chain = prompt | llm | StrOutputParser()

response = chain.invoke({
    "name": "Faisal",
    "education": "MCA",
    "skills": "Python, ML, DL, SQL, LangChain, RAG",
    "experience": "Fresher with internship experience",
    "career_goal": "Artificial Intelligence",
    "job_role": "AI/ML Engineer"
})

print(response)
```

# Program 2: AI Book Reviewer

```python

from langchain_core.prompts import ChatPromptTemplate

# Step 1: Create ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
        You are an expert book reviewer.
        You have extensive knowledge of literature and books.
        """
    ),
    (
        "human",
        """
        Book Name: {Book_Name}
        Author Name: {Author_Name}
        Category: {Category}

        Task:
        Review the given book in Urdu language.

        Include:
        1. Book Introduction
        2. Story Overview
        3. Main Characters
        4. Important Themes
        5. Strengths and Weaknesses
        6. Final Rating out of 10
        7. Overall Recommendation

        Write the entire review in Urdu.
        """
    )
])

# Step 2: Format Messages
result = prompt.format_messages(
    Book_Name="راجہ گدھ",
    Author_Name="بانو قدسیہ",
    Category="Philosophical Fiction"
)

# Step 3: Display Output
for message in result:
    print(message.type.upper())
    print(message.content)
    print("-" * 50)
```

```python

from langchain_core.output_parsers import StrOutputParser

# Step 1: Create chain
chain = prompt | llm | StrOutputParser()

# Step 2: Invoke chain
response = chain.invoke({
    "Book_Name": "راجہ گدھ",
    "Author_Name": "بانو قدسیہ",
    "Category": "Philosophical Fiction"
})

# Step 3: Print response
print(response)
```

# Program 3 : Customer Support

```python

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Step 1: Create ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
        You are an intelligent and professional Customer Support Assistant.

        Your responsibilities:

        1. Answer customer questions politely.
        2. Help customers with orders and delivery.
        3. Handle product-related queries.
        4. Assist with refund and cancellation requests.
        5. Resolve customer complaints.
        6. Provide clear and helpful solutions.
        7. Escalate complex problems to human support.

        Important Rules:

        - Always maintain a friendly and professional tone.
        - Never provide false information.
        - Never invent order status or refund confirmation.
        - If information is missing, ask the customer for it.
        - Keep answers simple and understandable.
        - If you cannot solve an issue, recommend human support.
        """
    ),
    (
        "human",
        """
        Customer Name: {customer_name}

        Customer Issue: {customer_issue}

        Order ID: {order_id}

        Product Name: {product_name}

        Preferred Language: {language}

        Based on the information provided:

        1. Understand the customer's problem.
        2. Identify the main issue.
        3. Provide a suitable solution.
        4. Give step-by-step instructions if required.
        5. Ask for additional information if necessary.

        Respond in {language}.
        """
    )
])

# Step 2: Create LCEL Chain

chain = prompt | llm | StrOutputParser()

# Step 3: Invoke the Chain

response = chain.invoke({
    "customer_name": "Faisal",
    "customer_issue": "I ordered a laptop 7 days ago, but it has not been delivered yet.",
    "order_id": "ORD12345",
    "product_name": "HP Laptop",
    "language": "English"
})

# Step 4: Display Response

print(response)
```

# Program 4 :Email Writing Assistant

```python

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Step 1: Create ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
        You are an intelligent and professional Email Writing Assistant.

        Your responsibilities:

        1. If the user wants to write an informal email to a friend
           or family member, use a friendly and natural tone.

        2. If the user wants to write a professional email to a boss,
           office, company, or institution, use a formal and
           professional tone.

        3. Emails should be neither too long nor too short.

        4. Maintain proper email formatting.

        5. Correct grammar and spelling mistakes.

        6. Suggest an appropriate subject if required.

        7. Do not change the original meaning of the user's message.

        8. Include an appropriate greeting and closing.

        9. Return only the final email.
        """
    ),
    (
        "human",
        """
        Write an email using the following information:

        Recipient: {email}
        Date: {date}
        Subject: {subject}
        Email Body / Main Message: {body}

        Generate a properly formatted email.
        """
    )
])

# Step 2: Create LCEL Chain

chain = prompt | llm | StrOutputParser()

# Step 3: Invoke the Chain

response = chain.invoke({
    "email": "My Professor",
    "date": "14 November 2026",
    "subject": "Request for Assignment Extension",
    "body": "I am unable to submit my assignment on time because I have some personal work. Please give me two extra days."
})

# Step 4: Print Response

print(response)
```

# Program 5:	Research Assistant	Multi-message prompts

```python

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Step 1: Create Multi-Message Prompt

prompt = ChatPromptTemplate.from_messages([

    # System Message
    (
        "system",
        """
        You are an expert AI Research Assistant.

        Your responsibilities:

        1. Analyze research topics deeply.
        2. Explain complex concepts in simple language.
        3. Identify important research findings.
        4. Discuss advantages and disadvantages.
        5. Identify research gaps.
        6. Suggest future research directions.
        7. Generate meaningful research questions.

        Rules:
        - Provide accurate and structured information.
        - Avoid inventing research papers, authors or citations.
        - Maintain an academic but understandable tone.
        - Follow the requested output language.
        """
    ),

    # Human Message
    (
        "human",
        """
        Research Topic: {topic}

        Research Field: {field}

        Research Level: {level}

        Output Language: {language}

        Research Requirements:
        1. Introduction to the topic.
        2. Detailed explanation.
        3. Important concepts and technologies.
        4. Advantages and disadvantages.
        5. Real-world applications.
        6. Current challenges.
        7. Research gaps.
        8. Five future research questions.

        Prepare a comprehensive research report.
        """
    ),

    # AI Message
    (
        "ai",
        """
        I will analyze the research topic systematically,
        explain its important concepts, and identify
        potential research opportunities.
        """
    ),

    # Additional Human Message
    (
        "human",
        """
        Also provide a concise executive summary
        and suggest relevant keywords for further study.
        """
    )

])

# Step 2: Create LCEL Chain

chain = prompt | llm | StrOutputParser()

# Step 3: Invoke the Chain

response = chain.invoke({

    "topic": "Retrieval Augmented Generation (RAG)",

    "field": "Artificial Intelligence",

    "level": "Advanced",

    "language": "English"

})

# Step 4: Print Research Report

print(response)
```
