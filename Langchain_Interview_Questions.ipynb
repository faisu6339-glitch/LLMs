# Question 3:Write a Python program using LangChain's PromptTemplate and your existing NVIDIA-backed llm.

Create a chat-based LLM chain
Beginner

Interview question

Write a program using ChatPromptTemplate to create an AI interview tutor.

Your program must:
1. Create a system message that defines the assistant's role as a technical interview tutor.
2. Create a human message containing two variables: {topic} and {difficulty}.
3. Connect the prompt to your existing llm using LCEL.
4. Use StrOutputParser() to return plain text.
5. Invoke the chain for two different topics and difficulty levels.

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
prompt=PromptTemplate(
    template="Explain me a {topic} at this {level}",
    input_variable=["topic","level"])

chain=prompt | llm | StrOutputParser()


result1=chain.invoke({
    "topic":"Machine Learning",
    "level":"Beginner"})

result2 = chain.invoke({
    "topic": "Transformer Architecture",
    "level": "advanced"
})

print("Answer 1: Machine Learning")
print(result1)

print("\nAnswer 2: Transformer Architecture")
print(result2)
```

Control LLM creativity
Beginner

Interview question

Write a Python program using your existing NVIDIA-backed LangChain model to compare responses generated with two different temperature settings: 0.1 and 0.9.

Your program must:
1. Create two model instances using the same NVIDIA model.
2. Set the first model's temperature to 0.1.
3. Set the second model's temperature to 0.9.
4. Send the same creative question to both models.
5. Print both responses with clear labels.
6. Explain why the answers might differ.
Use this question for both calls:

"Invent a creative name and a short description for an AI-powered robot that helps students learn."

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Step 1: Create the chat prompt
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an AI technical interview tutor. "
               "Explain concepts clearly and provide interview-focused "
               "explanations appropriate to the requested difficulty level."),

    ("human", "Explain {topic} at the {difficulty} level.")
])

# Step 2: Build the chain
chain = prompt | llm | StrOutputParser()

# Step 3: Invoke for the first topic
result1 = chain.invoke({
    "topic": "LangChain",
    "difficulty": "beginner"
})

print("=== LangChain: Beginner ===")
print(result1)

# Step 4: Invoke for the second topic
result2 = chain.invoke({
    "topic": "Self-Attention",
    "difficulty": "advanced"
})

print("\n=== Self-Attention: Advanced ===")
print(result2)
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

# Model with low temperature
llm1 = ChatOpenAI(
    model="nvidia/nemotron-3.5-lightning-30b-a3b",
    base_url="https://integrate.api.nvidia.com/v1",
    api_key=api_key,
    temperature=0.1,
    max_retries=3,
    timeout=60
)

# Model with high temperature
llm2 = ChatOpenAI(
    model="nvidia/nemotron-3.5-lightning-30b-a3b",
    base_url="https://integrate.api.nvidia.com/v1",
    api_key=api_key,
    temperature=0.9,
    max_retries=3,
    timeout=60
)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a creative AI assistant."),
    ("human", "{question}")
])

# Build two chains
chain1 = prompt | llm1 | StrOutputParser()
chain2 = prompt | llm2 | StrOutputParser()

question = (
    "Invent a creative name and a short description for an "
    "AI-powered robot that helps students learn."
)

result1 = chain1.invoke({"question": question})
result2 = chain2.invoke({"question": question})

print("=== Temperature 0.1 ===")
print(result1)

print("\n=== Temperature 0.9 ===")
print(result2)
```

# Question 6:— Zero-shot classification
Classify customer sentiment
Beginner

Interview question

Build an LLM chain that classifies a customer review into exactly one of these categories:
- Positive
- Negative
- Neutral
Requirements
1. Use your existing NVIDIA-backed llm.
2. Create a ChatPromptTemplate.
3. Use a system prompt that restricts the output to the three categories.
4. Connect the prompt, model, and StrOutputParser() using LCEL.
5. Test the chain with three reviews: one positive, one negative, and one neutral.
6. Print each review and its predicted sentiment.
Bonus: Explain why this is called zero-shot classification

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Step 1: Create the prompt
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a customer sentiment classifier.
        Classify the customer review into exactly one category:
        Positive, Negative, or Neutral.
        Return only the category name."""
    ),
    ("human", "{review}")
])

# Step 2: Build the chain
chain = prompt | llm | StrOutputParser()

# Step 3: Test three customer reviews
product_reviews = [
    "This phone is excellent and the battery lasts all day.",
    "This phone is terrible and stopped working after two days.",
    "The phone is black and has a 6.5-inch display."
]

# Step 4: Classify each review
for review in product_reviews:
    result = chain.invoke({"review": review})

    print("Review:", review)
    print("Sentiment:", result.strip())
    print("-" * 50)
```

#Question 7: — Text Summarization
Difficulty: Beginner

Interview problem

Build an LLM-based text summarizer using your existing NVIDIA API model and LangChain.

Your program must:
1. Create a ChatPromptTemplate with a system message and a human message.
2. Accept a long text through a variable named {text}.
3. Instruct the LLM to summarize the text into exactly 5 bullet points.
4. Build a chain using prompt | llm | StrOutputParser().
5. Invoke the chain with a sample paragraph.
6. Print the summary as plain text.

```python
from langchain_core.prompts import ChatPromptTemplate

# Step 1: Create the prompt
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a text summarization assistant. "
        "Summarize the given text into exactly 5 bullet points."
    ),
    ("human", "{text}")
])

# Step 2: Build the chain
chain = prompt | llm

# Step 3: Define the input text
input_text = """
Artificial Intelligence is transforming many industries.
In healthcare, AI helps doctors analyze medical images
and identify diseases. In education, AI-powered systems
provide personalized learning experiences. Businesses use
AI to automate repetitive tasks, improve customer service,
and analyze large amounts of data. However, AI also raises
concerns about privacy, bias, job displacement, and security.
Organizations must develop and use AI responsibly.
"""

# Step 4: Invoke the chain
response = chain.invoke({"text": input_text})

# Step 5: Extract the generated answer
summary = str(response.content)

# Step 6: Display the summary
print("===== TEXT SUMMARY =====")
print(summary)
```
